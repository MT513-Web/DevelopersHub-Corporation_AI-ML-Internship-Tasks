## Objective
I built this chatbot to help people get quick health tips and wellness advice. It uses Google's Gemini AI to understand questions and reply in a friendly, supportive way.

## Note:  It doesn’t replace a doctor.

Tools I Used: 
* Python (The language)

* Google GenAI SDK (To talk to Gemini)

* Prompt Engineering (To give the bot a doctor-like personality)

## Setting Up the Connection: 
Before the bot can talk, I need to connect my code to Google's AI server. I used my unique API Key to get access. Think of this as the "Passport" for my code to travel to the AI world.

```python
from google import genai
import os
from dotenv import load_dotenv

# Load our secret keys from the .env file
load_dotenv()

# Grab the API key securely
api_key = os.getenv("GOOGLE_API_KEY")

# Initialize the Gemini client the modern way
client = genai.Client(api_key=api_key)

# Let's check if we're connected by listing the models
print("Connection successful! Available Models:")
for m in client.models.list():
    print(f"-> {m.name}")
```

### Setting up the Health Persona

Training the Bot to be a "Doctor":
  Standard AI can talk about anything (movies, sports, etc.). But for this project, I gave it strict rules. I told it: "You are a Health Assistant, be kind, and always tell users to consult be real doctor for medicines." This makes the bot safer and smarter.

```python
health_persona = """
You are a professional and friendly Health Assistant. 
Your goal is to provide helpful health and wellness advice. 
Rules:
1. Always start your conversation with 'Assalam-o-Alaikum'.
2. Always give a disclaimer that you are an AI, not a real doctor.
3. If someone asks for a specific medicine dose, tell them to consult a professional.
4. Keep your tone supportive and encouraging.
"""

# Now we update our request to include this persona
response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=f"Context: {health_persona} \n\n User Question: Give me a health tip for someone who sits all day at a desk."
)

print(response.text)
```

### 3. Smart Health Chat Function
Creating the Brain (Function):
 I created a simple function that takes a user's question, mixes it with our "Doctor Rules," and sends it to the AI. This keeps the code clean and reusable.

```python
# This function makes it easy to ask health questions
def ask_health_bot(user_question):
    # We combine our doctor rules with the user's question
    full_prompt = f"Context: {health_persona} \n\n Question: {user_question}"
    
    # Send the request to Gemini
    response = client.models.generate_content(
        model="gemini-flash-latest",
        contents=full_prompt
    )
    
    # Return only the text answer
    return response.text
```

### 4. Live Interactive Health Chat

Live Chat Session:
 This is where I actually test the bot! I created a loop so I can have a real conversation. If I type 'exit', the conversation ends gracefully.

```python
# This loop lets you chat with the bot live
print("--- Health Bot is Online! (Type 'exit' to stop) ---")

while True:
    # Get the question from us
    user_input = input("You: ")
    
    # Check if we want to stop
    if user_input.lower() == 'exit':
        print("Bot: Allah Hafiz! Stay healthy.")
        break
    
    # Get the answer using our smart function
    bot_reply = ask_health_bot(user_input)
    
    # Print the answer
    print(f"Bot: {bot_reply}")
    print("-" * 30)
```
